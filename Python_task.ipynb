```python
#
numbers = [20, 1, 58, 45, 33]

largest = second_largest = float('-inf')

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print(second_largest)
```

```python
for prime in range(2, 101):
    is_prime = True

    for i in range(2, prime):
        if prime % i == 0:
            is_prime = False
            break

    if is_prime:
        print(prime, end=' ')
```

```python
def fibonacci(n):
    a = 0
    b = 1

    for i in range(n):
        print(a, end=" ")
        c = a + b
        a = b
        b = c

n = int(input("Enter number: "))
fibonacci(n)
```

```python
text = input("Enter a string: ")

digits = 0
alphabets = 0
special = 0

for char in text:
    if char.isdigit():
        digits += 1
    elif char.isalpha():
        alphabets += 1
    elif not char.isspace():
        special += 1

print("Total Digits:", digits)
print("Total Alphabets:", alphabets)
print("Total Special Characters:", special)
```
